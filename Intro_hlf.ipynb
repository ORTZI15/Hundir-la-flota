## Crear barcos (TERMINADA)

- 3 barcos de eslora 2.
- 2 de eslora 3.
- 1 eslora 4.

```python
def crear_barco(eslora):
    barco = []
    pos_ini = (np.random.randint(10), np.random.randint(10))
    barco.append(pos_ini)
    dir = np.random.choice(("V", "H"))
    p = 1
    row, col = pos_ini

    if dir == "V" and (row + eslora) <= 10:
        for pos in range(eslora - 1):
            barco.append((pos_ini[0] + p, pos_ini[1]))
            p = p + 1
    elif dir == "H" and (col + eslora) <= 10:
        for pos in range(eslora - 1):
            barco.append((pos_ini[0], pos_ini[1] + p))
            p = p + 1   
    else:
        print("calculando barco de nuevo...")
        barco = crear_barco(eslora)

    return barco
```

```python
print(crear_barco(2))
print(crear_barco(4))
```

## Crear flota (TERMINADA)

```python
def crear_flota():
    esloras_flota = [4, 3, 3, 2, 2, 2]
    flota = []
    for eslora in esloras_flota:
        flota.append(crear_barco(eslora))

    lista_posiciones = []
    for barco in flota:
        for pos in barco:
            lista_posiciones.append(pos)

    if len(lista_posiciones) != len(set(lista_posiciones)):
        flota = crear_flota()
    
    return flota
```
