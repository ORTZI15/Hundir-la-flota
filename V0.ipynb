```python
import numpy as np  # Importamos NumPy para generar valores aleatorios

tamano_tablero = 10  # Definimos el tamaño del tablero
barcos = [4, 3, 3, 2, 2, 2]  # Definimos las longitudes de los barcos

def colocar_barcos():  # Definimos la función de colocación
    for longitud in barcos:  # Recorremos cada barco
        fila = np.random.randint(0, tamano_tablero)  # Elegimos una fila aleatoria
        columna = np.random.randint(0, tamano_tablero)  # Elegimos una columna aleatoria
        orientacion = np.random.choice(["horizontal", "vertical"])  # Elegimos la orientación

        if orientacion == "horizontal":  # Comprobamos si el barco es horizontal
            cabe = columna + longitud <= tamano_tablero  # Verificamos el límite horizontal
        else:  # Si no es horizontal, es vertical
            cabe = fila + longitud <= tamano_tablero  # Verificamos el límite vertical

        print(longitud, fila, columna, orientacion, cabe)  # Mostramos los datos y el resultado

colocar_barcos()  # Ejecutamos la función
```
